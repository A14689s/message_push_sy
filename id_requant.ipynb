```python
import requests

TOKEN = "token"

proxies = {
    "http":  "",
    "https": "",
}

url = f"https://api.telegram.org/bot{TOKEN}/getMe"
r = requests.get(url, proxies=proxies, timeout=10)
print(r.status_code, r.text)

```
