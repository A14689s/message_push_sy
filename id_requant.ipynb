```python
import asyncio
import python_socks
import nest_asyncio
from telethon import TelegramClient, events

# 1. 解决 Jupyter/Anaconda 环境下的 asyncio.run() 冲突
nest_asyncio.apply()

# --- 配置区 ---
TG_API_ID=
TG_API_HASH=""
BOT_TOKEN = '' # 从 @BotFather 获取的 Token

# 代理设置: 使用 HTTP 协议和 10809 端口
PROXY = (python_socks.ProxyType.HTTP, '127.0.0.1', 10809)

# 初始化客户端 (使用 Bot Token 登录)
# session 文件名设为 'test_bot'，运行后本地会生成 test_bot.session
client = TelegramClient('test_bot', TG_API_ID, TG_API_HASH, proxy=PROXY)

# --- 自动回复逻辑 ---
@client.on(events.NewMessage(incoming=True, func=lambda e: e.is_private))
async def handle_private_message(event):
    """当有人给机器人发私信时触发"""
    sender = await event.get_sender()
    sender_name = sender.first_name if sender else "未知用户"
    
    print(f"📩 收到消息: [{sender_name}] (ID: {event.sender_id}) -> {event.text}")
    
    # 自动回复
    response_text = f"✅ 测试通过！\n你的 ID 是: {event.sender_id}\n机器人已收到消息: {event.text}"
    await event.reply(response_text)
    
    print(f"🚀 已自动回复给 {sender_name}")

# --- 主运行程序 ---
async def main():
    print("🚀 正在通过代理 10809 连接 Telegram...")
    try:
        # 使用 Token 启动
        await client.start(bot_token=BOT_TOKEN)
        
        me = await client.get_me()
        print(f"🤖 机器人 @{me.username} 已上线！")
        print("📢 提示：现在请用你的手机 Telegram 给这个机器人发条消息进行测试。")
        
        # 保持运行，直到手动停止
        await client.run_until_disconnected()
        
    except Exception as e:
        print(f"❌ 发生错误: {e}")

# 执行启动
if __name__ == '__main__':
    # 有了 nest_asyncio，这里可以直接调用 asyncio.run()
    asyncio.run(main())
```
